```python
import os
current_dir = os.getcwd()
parent_dir = os.path.dirname(current_dir)

os.chdir(parent_dir)

import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
```

```python
def plot_adj_heatmap(matrix, title, cmap='magma', annot=False):
    plt.figure(figsize=(10, 8))
    
    sns.heatmap(
        matrix, 
        cmap=cmap, 
        cbar=True, 
        annot=annot, 
        fmt=".2f",
        xticklabels=False,
        yticklabels=False,
        cbar_kws={'label': 'Connection Strength'}
    )
    
    plt.title(title, fontsize=15, fontweight='bold', pad=20)
    plt.xlabel("Node Index", fontsize=12)
    plt.ylabel("Node Index", fontsize=12)
    
    plt.tight_layout()
    plt.show()

def optimize_pems_adj(dist_path, sigma_scale=5.0, epsilon=0.05):
    dist_matrix = pd.read_csv(dist_path, header=None).values
    std_val = np.std(dist_matrix[dist_matrix > 0])
    sigma = std_val * sigma_scale
    
    W = np.exp(-(dist_matrix**2) / (sigma**2))
    W[W < epsilon] = 0
    np.fill_diagonal(W, 0)
    return W

W_228 = optimize_pems_adj('./dataset/PeMSD7/PeMSD7_Full/PeMSD7_W_228.csv', sigma_scale=5, epsilon=0.05)
plot_adj_heatmap(W_228, "Adjacency Matrix: PeMSD7 (M, 228 Nodes)")

W_1026 = optimize_pems_adj('./dataset/PeMSD7/PeMSD7_Full/PeMSD7_W_1026.csv', sigma_scale=5, epsilon=0.05)
plot_adj_heatmap(W_1026, "Adjacency Matrix: PeMSD7 (L, 1026 Nodes)")

W_shanghai = pd.read_csv('./dataset/ShanghaiRailway/W_43.csv', header=None).values
plot_adj_heatmap(W_shanghai, "Adjacency Matrix: Shanghai Railway (43 Nodes)")

W_air_adj = np.load('./dataset/Airquality/air_quality_adj.npy')
plot_adj_heatmap(
    W_air_adj, 
    "Adjacency Matrix: Air Quality (12 Stations)",
    annot=True
)
```
